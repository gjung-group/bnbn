```python
import numpy as np
import math
import matplotlib.pyplot as plt
def alpha(a,b,ap,bp):#Herman equation 29
    p = 1
    gamma = a**2 + b ** 2 + a*b #equation 27
    alpha_rad = np.arccos((2*ap*a + 2*bp*b+ap*b+bp*a)/(2*p*gamma))
    alpha_deg = (alpha_rad*180/np.pi)
    return alpha_deg
angle=[]
for i in range(30):
    angle.append(alpha(i+1,i+2,i+2,i+1))
angle=np.array(angle)
aa_re=np.loadtxt('aa_relax')
aa_ri=np.loadtxt('aa_rigid')
aap_re=np.loadtxt('aap_relax')
aap_rigid=np.loadtxt('aap_rigid')

aa_ri=np.loadtxt('aa_rigid')
plt.figure(figsize=(13,6))
plt.subplot(121)
plt.plot(angle,aa_re[:,0],'o-',label='AA_rel',c='maroon')#AA
plt.plot(angle,aa_re[:,1],'o-',label='AB_rel',c='gold')#AB
plt.plot(angle,aa_re[:,2],'o-',label='BA_rel',c='blue',alpha=0.6)#BA
# plt.plot(angle,aa_re[:,3],'o-',label='DW_re',c='navy')#DW
plt.scatter(angle,aa_ri[:,0],label="AA_rig",marker='*',c='maroon')#AA
plt.scatter(angle,aa_ri[:,1],label="AB_rig",marker='*',c='gold')#AB
plt.scatter(angle,aa_ri[:,2],label="BA_rig",marker='*',c='blue',alpha=0.5)#BA
# plt.scatter(angle,aa_ri[:,3],label="DW'_ri",marker='*',c='grey')#DW
plt.ylabel('Stacking area(%)',fontsize=22)
plt.xlabel('$\Theta$' '($^{\circ}$)',fontsize=22)
plt.yticks(fontsize=21)
plt.xticks(fontsize=21)
plt.xlim(1,8)
plt.legend(loc='best',fontsize=15)
plt.tight_layout()
plt.title('BN/BN',fontsize=22)
plt.subplot(122)
plt.plot(angle,aap_re[:,0],'o-',label="AA'_rel",c='maroon')#AA
plt.plot(angle,aap_re[:,1],'o-',label="AB'_rel",c='gold')#AB
plt.plot(angle,aap_re[:,2],'o-',label="BA'_rel",c='blue',alpha=0.6)#BA
# plt.plot(angle,aap_re[:,3],'o-',label="DW'_re",c='navy')#DW
plt.scatter(angle,aap_rigid[:,0],label="AA'_rig",marker='*',c='maroon')#AA
plt.scatter(angle,aap_rigid[:,2],label="BA'_rig",marker='*',c='blue')#BA
plt.scatter(angle,aap_rigid[:,1],label="AB'_rig",marker='*',c='gold')#AB

# plt.scatter(angle,aap_rigid[:,3],label="DW'_ri",marker='*',c='grey')#DW
plt.ylabel('stacking area (%)',fontsize=22)
plt.yticks(fontsize=21)
plt.xticks(fontsize=21)
plt.xlim(1,8)
plt.xlabel('twist angle',fontsize=22)
plt.xlabel('$\Theta$' '($^{\circ}$)',fontsize=22)
plt.title('BN/NB',fontsize=22)
plt.legend(loc='best',fontsize=15)
plt.tight_layout()
plt.savefig('domain',dpi=600)
```
