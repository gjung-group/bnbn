```python
import numpy as np
import matplotlib.pyplot as plt
x=np.loadtxt('Re_AAp.dat',usecols=(0,),delimiter=",").reshape(11,19)
y=np.loadtxt('Re_AAp.dat',usecols=(1,),delimiter=",").reshape(11,19)
np.savetxt('xaxis',x)
np.savetxt('yaxis',y)

z13=np.loadtxt('Re_AAp.dat',usecols=(2,),delimiter=",").reshape(11,19)
z24=np.loadtxt('Re_BBp.dat',usecols=(2,),delimiter=",").reshape(11,19)
z14=np.loadtxt('Re_BAp.dat',usecols=(2,),delimiter=",").reshape(11,19)
z23=np.loadtxt('Re_ABp.dat',usecols=(2,),delimiter=",").reshape(11,19)
np.savetxt('AB-REAL-dft',z23)


TC13=np.loadtxt('TC-13-real',usecols=(0,)).reshape(11,19)
TC24=np.loadtxt('TC-24-real',usecols=(0,)).reshape(11,19)
TC23=np.loadtxt('TC-23-real',usecols=(0,)).reshape(11,19)
TC14=np.loadtxt('TC-14-real',usecols=(0,)).reshape(11,19)
np.savetxt('AB-REAL-TC',TC23)

DIF13=(z13-TC13)
DIF24=(z24-TC24)
DIF23=(z23-TC23)
DIF14=(z14-TC14)

np.savetxt('DIFF-REAL',DIF23)

plt.figure(figsize=(15,12))
plt.subplot(341)
plt.contourf(x,y,z13,44,cmap='Blues')
plt.colorbar()
plt.title("DFT-real-AA'")

plt.subplot(342)
plt.contourf(x,y,z24,44,cmap='Blues')
plt.colorbar()
plt.title("DFT-real-BB'")

plt.subplot(343)
plt.contourf(x,y,z23,44,cmap='Blues')
plt.colorbar()
plt.title("DFT-real-AB'")

plt.subplot(344)
plt.contourf(x,y,z14,44,cmap='Blues')
plt.colorbar()
plt.title("DFT-real-BA'")

plt.subplot(345)
plt.contourf(x,y,TC13,44,cmap='Reds')
plt.colorbar()
plt.title("TC-real-AA'")

plt.subplot(346)
plt.contourf(x,y,TC24,44,cmap='Reds')
plt.colorbar()
plt.title("TC-real-BB'")

plt.subplot(347)
plt.contourf(x,y,TC23,44,cmap='Reds')
plt.colorbar()
plt.title("TC-real-AB'")

plt.subplot(348)
plt.contourf(x,y,TC14,44,cmap='Reds')
plt.colorbar()
plt.title("TC-real-BA'")

plt.subplot(349)
plt.contourf(x,y,DIF13,44,cmap='RdBu')
plt.colorbar()
plt.title("Difference-real-AA'")

plt.subplot(3,4,10)
plt.contourf(x,y,DIF24,44,cmap='RdBu')
plt.colorbar()
plt.title("Difference-real-BB'")

plt.subplot(3,4,11)
plt.contourf(x,y,DIF23,44,cmap='RdBu')
plt.colorbar()
plt.title("Difference-real-AB'")

plt.subplot(3,4,12)
plt.contourf(x,y,DIF14,44,cmap='RdBu')
plt.colorbar()
plt.title("Difference-real-BA'")

plt.tight_layout()
print (np.min(DIF13)/np.max(z13))
print (np.min(DIF24)/np.max(z24))
print (np.min(DIF23)/np.max(z23))
```
